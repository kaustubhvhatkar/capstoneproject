```python
!git clone https://github.com/kaustubhvhatkar/capstoneproject.git

import numpy as np
import os

REPO_DIR = '/content/capstoneproject'   # change to your repo folder name

data = {}   # will hold data['function_1'] = {'inputs': ..., 'outputs': ...}

for i in range(1, 9):
    func_name = f'function_{i}'
    folder = os.path.join(REPO_DIR, func_name)

    inputs_path  = os.path.join(folder, 'initial_inputs.npy')
    outputs_path = os.path.join(folder, 'initial_outputs.npy')

    # Quick check the files actually exist before loading
    if not (os.path.exists(inputs_path) and os.path.exists(outputs_path)):
        print(f'⚠️  Missing files in {func_name}, skipping')
        continue

    data[func_name] = {
        'inputs':  np.load(inputs_path),
        'outputs': np.load(outputs_path),
    }

    print(f'{func_name}: inputs {data[func_name]["inputs"].shape}, '
          f'outputs {data[func_name]["outputs"].shape}')

print(f'\nLoaded {len(data)} functions total')




for func_name, d in data.items():
    y = d['outputs']
    X = d['inputs']

    # argmax gives the index of the highest output
    best_idx = np.argmax(y)

    best_input  = X[best_idx]
    best_output = y[best_idx]

    print(f'{func_name}:')
    print(f'  Best index : {best_idx}')
    print(f'  Best input : {best_input}')
    print(f'  Best output: {best_output}')
    print()
```
