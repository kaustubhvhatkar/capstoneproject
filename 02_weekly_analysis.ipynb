```python
import pandas as pd
import numpy as np

from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns
import ast

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)


def parse_array_file(path):
    """Parse files containing either a Python literal or array(...) calls."""
    text = path.read_text().strip()

    # Fast path: pure Python literal
    try:
        return np.asarray(ast.literal_eval(text)).tolist()
    except (ValueError, SyntaxError):
        pass

    # Fallback: file contains array(...) calls
    safe = {"array": np.array, "np": np, "numpy": np}
    val = eval(f"[{text}]", {"__builtins__": {}}, safe)
    return np.asarray(val).tolist()


ROOT = Path("/content/capstoneproject/submissions")

records = []

for week_dir in sorted(ROOT.glob("week*")):

    week_num = int(week_dir.name.replace("week", ""))

    inputs_file = week_dir / "inputs.txt"
    outputs_file = week_dir / "outputs.txt"

    if not inputs_file.exists() or not outputs_file.exists():
        continue

    try:
        queries = parse_array_file(inputs_file)
        results = parse_array_file(outputs_file)
    except Exception as e:
        print(f"Skipping {week_dir.name}: {e}")
        continue

    for idx, (query, result) in enumerate(zip(queries, results), start=1):

        records.append({
            "week": week_num,
            "function": f"function_{idx}",
            "query": query,
            "result": float(np.asarray(result).squeeze())
        })

df = pd.DataFrame(records)

print(f"Loaded {len(df)} observations")
df.head()
```
